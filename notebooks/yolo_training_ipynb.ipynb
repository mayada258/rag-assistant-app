```python
import os
import shutil
import random
from collections import defaultdict

random.seed(42)

images_dir = "/content/dataset_combined/images"
labels_dir = "/content/dataset_combined/labels"

class_names = ['battery', 'electronic-waste', 'keyboard', 'laptop', 'mobile phone',
               'monitor', 'mouse', 'pcb', 'printer']

label_files = [f for f in os.listdir(labels_dir) if f.endswith(".txt")]

class_to_files = defaultdict(list)
for lf in label_files:
    with open(os.path.join(labels_dir, lf)) as f:
        classes_in_file = set(int(line.split()[0]) for line in f if line.strip())
    if not classes_in_file:
        continue
    rarest_class = min(classes_in_file, key=lambda c: len(class_to_files[c]) if c in class_to_files else 0)
    class_to_files[rarest_class].append(lf)

train_files = set()
test_files = set()

for cls, files in class_to_files.items():
    random.shuffle(files)
    split_idx = int(len(files) * 0.8)
    train_files.update(files[:split_idx])
    test_files.update(files[split_idx:])

print("Number of train files:", len(train_files))
print("Number of test files:", len(test_files))
```

```python
new_base = "/content/dataset_v2"
new_train_img = os.path.join(new_base, "train/images")
new_train_lbl = os.path.join(new_base, "train/labels")
new_test_img = os.path.join(new_base, "test/images")
new_test_lbl = os.path.join(new_base, "test/labels")

for d in [new_train_img, new_train_lbl, new_test_img, new_test_lbl]:
    os.makedirs(d, exist_ok=True)

def get_image_filename(label_filename):
    base_name = os.path.splitext(label_filename)[0]
    for ext in [".jpg", ".jpeg", ".png"]:
        candidate = base_name + ext
        if os.path.exists(os.path.join(images_dir, candidate)):
            return candidate
    return None

for lf in train_files:
    img_file = get_image_filename(lf)
    if img_file:
        shutil.copy(os.path.join(images_dir, img_file), new_train_img)
        shutil.copy(os.path.join(labels_dir, lf), new_train_lbl)

for lf in test_files:
    img_file = get_image_filename(lf)
    if img_file:
        shutil.copy(os.path.join(images_dir, img_file), new_test_img)
        shutil.copy(os.path.join(labels_dir, lf), new_test_lbl)

print("Train images copied:", len(os.listdir(new_train_img)))
print("Test images copied:", len(os.listdir(new_test_img)))
```

## 2.6 Evaluation — YOLO Vision Component

### Initial Split (v1)
The first train/test split was a simple random split, which resulted in **2 out of 9 classes (`pcb`, `printer`) having zero instances in the test set** — despite being present in training (479 and 185 instances respectively). This made it impossible to fairly evaluate the model on these classes.

Initial results (random split):
- mAP50: 0.399
- mAP50-95: 0.282
- Several classes (`battery`, `mouse`, `electronic-waste`) scored very low (mAP50 < 0.12), largely due to insufficient/unbalanced test representation.

### Fix: Stratified Split (v2)
To address this, a stratified split was implemented: each image was assigned to a class bucket based on its rarest annotated class, then split 80/20 within each bucket. This ensured all 9 classes were represented in both train and test sets.

### Final Results (v2, YOLOv8n, 25 epochs)

| Class | Images | Instances | Precision | Recall | mAP50 | mAP50-95 |
|---|---|---|---|---|---|---|
| all | 1408 | 1412 | 0.910 | 0.869 | 0.920 | 0.663 |
| battery | 258 | 258 | 0.908 | 0.888 | 0.916 | 0.721 |
| electronic-waste | 191 | 191 | 0.853 | 0.788 | 0.893 | 0.702 |
| keyboard | 114 | 114 | 0.964 | 0.947 | 0.964 | 0.637 |
| laptop | 159 | 162 | 0.954 | 0.887 | 0.952 | 0.584 |
| mobile phone | 285 | 285 | 0.979 | 0.947 | 0.984 | 0.735 |
| monitor | 129 | 129 | 0.912 | 0.891 | 0.951 | 0.686 |
| mouse | 140 | 140 | 0.937 | 0.952 | 0.979 | 0.690 |
| pcb | 96 | 96 | 0.779 | 0.740 | 0.748 | 0.551 |
| printer | 37 | 37 | 0.906 | 0.778 | 0.893 | 0.657 |

### Failure Case Analysis
The stratified split alone improved overall mAP50 from **0.399 to 0.920** (+0.521) and mAP50-95 from **0.282 to 0.663**, confirming the original bottleneck was data split quality, not model capacity.

The weakest remaining class is `pcb` (mAP50 0.748), likely due to:
- Smallest instance count among all classes (96, vs. 258+ for top classes)
- PCBs vary heavily in visual appearance (size, color, component layout) compared to more visually consistent classes like `monitor` or `mobile phone`

`laptop` also lags slightly on mAP50-95 (0.584), suggesting looser bounding-box precision — possibly due to laptops often being partially occluded or photographed at varying angles.

**Mitigation applied:** stratified splitting to guarantee class representation across train/test.
**Future improvement (not applied due to time constraints):** additional augmentation or oversampling for `pcb` and `printer`, or more training epochs.
